```python
import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"[^a-z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_title"] = df["title"].apply(clean_text)
df[["title", "clean_title"]].head()
```

```python
trusted_sources = [
    "bbc", "reuters", "theguardian", "nytimes", "cnn","economictimes.indiatimes"
]

def label_from_hostname(hostname):
    hostname = str(hostname).lower()
    for src in trusted_sources:
        if src in hostname:   # partial match
            return 0          # Real (trusted)
    return 1                  # Fake / less credible

df["label"] = df["hostname"].apply(label_from_hostname)

df[["hostname", "label"]].value_counts().head(10)
```
